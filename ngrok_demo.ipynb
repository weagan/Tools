```python
!pip install -q pyngrok streamlit
from pyngrok import ngrok
from google.colab import userdata

# Authenticate with your secret NGROK_AUTH_TOKEN
ngrok.set_auth_token(userdata.get('NGROK_AUTH_TOKEN'))
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np

st.title("🚀 Colab + ngrok GUI Demo")
st.write("This interface is running live from a Google Colab notebook!")

# Add an interactive slider
val = st.slider("Select a value", 0, 100, 50)
st.write(f"The current value is: **{val}**")

# Display some random data
chart_data = pd.DataFrame(np.random.randn(20, 3), columns=['a', 'b', 'c'])
st.line_chart(chart_data)
```

```python
import os

# 1. Start Streamlit in the background
os.system("nohup streamlit run app.py &")

# 2. Open the ngrok tunnel
try:
    public_url = ngrok.connect(8501).public_url
    print(f"✅ GUI is live at: {public_url}")
except Exception as e:
    print(f"Error: {e}")
```
